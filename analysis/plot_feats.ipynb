## DeepPlant as the baseline

`TF_targets_deep_plant.txt` has no score column, but its **line order is the ranking**
— SDG8, PolII, MED12, SYD, GRF3 sit at the top, which is not alphabetical and not the
order of any other file. Confirmed quantitatively: Spearman against the line order is
**positive for all five** other methods (+0.16 to +0.29), which would not happen for an
arbitrary ordering.

So DeepPlant is a sixth ranking, and the natural framing is to treat it as the
*baseline*: sort rows by DeepPlant rank and ask how well each method reproduces it.

```python
METHODS_DNA
```

```python
import os

# style tokens: sequential single hue, light -> dark (not a rainbow)
BLUE = LinearSegmentedColormap.from_list(
    "seqblue", ["#f2f6fb", "#c6dbef", "#7fb0d8", "#3d7fb8", "#12518c", "#08306b"])
INK, MUTED = "#1a1a1a", "#6b6b6b"
```

```python


odir = "../results/method_comparison_3"
os.makedirs(odir, exist_ok=True)

BASE = "ChromBERT ES"
TOP_N_DP = 20

def load_rankings_with_baseline(level="factor", feature_type="histone"):
    """As load_rankings, but with DeepPlant prepended as a sixth column.

    DeepPlant rank = position in the file (1 = first line). The score is negated
    so that, like every other column, higher score = more important.
    """
    if feature_type == "histone":
        methods, nmap = METHODS, name_map
    else:
        methods, nmap = METHODS_DNA, name_map_dna

    cols = {}
    for label, df, idc, sc in methods:
        print(label)
        
        d = df[[idc, sc]].copy()
        d.columns = ["id", "score"]
        if level == "factor":
            d["id"] = d["id"].map(nmap)
            d = d.dropna(subset=["id"]).groupby("id", as_index=False)["score"].max()
        cols[label] = pd.Series(d["score"].rank(ascending=False, method="min").values,
                                index=d["id"].values)
    
    return pd.DataFrame(cols).dropna(subset=[m[0] for m in methods])


rank6 = load_rankings_with_baseline("factor", "histone")
rank6_dna = load_rankings_with_baseline("factor", "dna")
print(f"{len(rank6)} factors | no DeepPlant rank for {int(rank6[BASE].isna().sum())}")
print(f"{len(rank6_dna)} DNA binding proteins | no DeepPlant rank for {int(rank6_dna[BASE].isna().sum())}")
print("\nSpearman vs DeepPlant (DNA binding proteins):")
for label, *_ in METHODS_DNA:
    print(f"  {label:24s} {spearmanr(rank6_dna[BASE], rank6_dna[label], nan_policy='omit').statistic:+.3f}")
print("\nSpearman vs DeepPlant (histone modifications):")
for label, *_ in METHODS:
    print(f"  {label:24s} {spearmanr(rank6[BASE], rank6[label], nan_policy='omit').statistic:+.3f}")
```

```python
plt.style.use("seaborn-v0_8-white")
TOP_N_DP = 15
def plot_dp_baseline(rank_df, n=TOP_N_DP, level="factor", rows="baseline", outfile=None):
    """Rank heatmap with DeepPlant as the leftmost baseline column.

    rows="baseline" -> DeepPlant's own top-n, in DeepPlant order
    rows="union"    -> union of all six top-n lists, still sorted by DeepPlant rank
    """
    methods = list(rank_df.columns)
    N = len(rank_df)
    tops = {m: set(rank_df[m].nsmallest(n).index) for m in methods}
    plot_methods = methods[1:]
    if rows == "baseline":
        feats = list(rank_df[BASE].nsmallest(n).index)
        sub = f"rows = DeepPlant's top {n}, in DeepPlant order"
    else:
        feats = list(rank_df.loc[sorted(set().union(*tops.values()))].sort_values(BASE).index)
        sub = f"rows = union of all 6 top-{n} lists, sorted by DeepPlant rank"

    mat = rank_df.loc[feats, plot_methods].values.astype(float)

    fig, ax = plt.subplots(figsize=(9.2, max(5.5, 0.36 * len(feats) + 2.6)))
    ax.imshow(mat, cmap=BLUE.reversed(), norm=LogNorm(vmin=1, vmax=N), aspect=0.4)

    for i, f in enumerate(feats):
        for j, m in enumerate(plot_methods):
            v = mat[i, j]
            ax.add_patch(Rectangle((j - .5, i - .5), 1, 1, fill=False,
                                   edgecolor="white", lw=2, zorder=2))
            if np.isnan(v):                       # in our set but absent from DeepPlant
                ax.add_patch(Rectangle((j - .5, i - .5), 1, 1, facecolor="#ededed",
                                       edgecolor="white", lw=2, zorder=3))
                ax.text(j, i, "–", ha="center", va="center", color=MUTED,
                        fontsize=8, zorder=5)
                continue
            if f in tops[m]:
                ax.add_patch(Rectangle((j - .42, i - .42), .84, .84, fill=False,
                                       edgecolor="#d94801", lw=2.0, zorder=4))
            ax.text(j, i, int(v), ha="center", va="center", zorder=5, fontsize=10,
                    color="white" if v <= 20 else MUTED,
                    fontweight="bold" if f in tops[m] else "normal")

    #ax.axvline(0.5, color=INK, lw=1.6, zorder=6)   # baseline | methods being judged

    # rho = {m: spearmanr(rank_df[BASE], rank_df[m], nan_policy="omit").statistic
    #        for m in methods[1:]}
    ax.set_xticks(range(len(plot_methods)))
    ax.set_xticklabels([f"{m}\n"  for m in plot_methods], 
                       fontsize=12, color=INK, rotation=45) #+ r"$\rho$=" + f"{rho[m]:+.2f}"
    ax.set_yticks(range(len(feats)))
    ax.set_yticklabels(feats, fontsize=12, color=INK)
    ax.tick_params(length=0)
    for s in ax.spines.values():
        s.set_visible(False)

    # ax.set_title(f"Comparison of different feature importance methods with top-20 in {BASE}",
    #              fontsize=12.5, color=INK, loc="left", fontweight="bold", pad = 20)
    # ax.text(0, 1.0, f"{sub}\ncell = rank of {N}, darker = more important   ·   "
    #                 f"orange ring = in that method's own top {n}   ·   "
    #                ,
    #         transform=ax.transAxes, fontsize=8.2, color=MUTED, va="bottom",
    #         linespacing=1.5)
    # r"$\rho$ = Spearman vs DeepPlant, all " + f"{N} {level}s"
    sm = plt.cm.ScalarMappable(cmap=BLUE.reversed(), norm=LogNorm(vmin=1, vmax=N))
    cb = fig.colorbar(sm, ax=ax, fraction=.030, pad=.02)
    cb.set_label(f"rank (log scale, 1 = top of {N})", fontsize=10, color=MUTED)
    cb.ax.tick_params(labelsize=10, length=3, color=MUTED)
    cb.outline.set_visible(False)

    if outfile:
        fig.savefig(outfile, dpi=600, bbox_inches="tight", facecolor="white")
    return fig


plot_dp_baseline(rank6, rows="baseline",
                 outfile=f"{odir}/deepplant_baseline_top{TOP_N_DP}_histone.png")
plot_dp_baseline(rank6, rows="union",
                 outfile=f"{odir}/deepplant_baseline_union_histone.png")
plt.show()
```

```python
# side-by-side per-gene / per-experiment PCC boxplots, with paired Wilcoxon signed-rank
# p-values of XGB vs every other method (pairs matched by gene / experiment index)
from scipy.stats import wilcoxon

def dp_by_index(df, idx_col, n):
    # DeepPlant files are sorted by PCC, so put them back in index order
    return df.set_index(idx_col).loc[range(n), "Pearsonr correlation"].values

n_exp, n_gene = len(xgb), len(xgb_df)

exp_pcc = {
    "ChromBERT": chrombert_exp["pcc"].values,
    "DeepPlant": dp_by_index(deepplant, "Experiment ID", n_exp),
    "RF": rf["pcc"].values,
    "MLP": mlp["pcc"].values,
    "XGBoost": xgb["pcc"].values,
}
gene_pcc = {
    "ChromBERT": chrombert_gene["pcc"].values,
    "DeepPlant": dp_by_index(deepplant_gene, "Gene index", n_gene),
    "RF": rf_gene["pcc"].values,
    "MLP": mlp_df["pcc"].values,
    "XGBoost": xgb_df["pcc"].values,
}
for name, pcc, n in [("experiment", exp_pcc, n_exp), ("gene", gene_pcc, n_gene)]:
    assert all(len(v) == n for v in pcc.values()), f"per-{name} lengths differ: { {k: len(v) for k, v in pcc.items()} }"


def xgb_wilcoxon(pcc, level):
    """Two-sided paired Wilcoxon signed-rank test of XGB vs each other method (NaN pairs dropped)."""
    ref = np.asarray(pcc["XGBoost"], dtype=float)
    pvals = {}
    print(f"\nPer-{level} PCC: XGB vs others (Wilcoxon signed-rank, two-sided)")
    for name, vals in pcc.items():
        if name == "XGBoost":
            continue
        vals = np.asarray(vals, dtype=float)
        ok = ~np.isnan(ref) & ~np.isnan(vals)
        res = wilcoxon(ref[ok], vals[ok])
        pvals[name] = res.pvalue
        print(f"  XGB vs {name:10s} n={ok.sum():5d}  median diff={np.median(ref[ok] - vals[ok]):+.4f}  p={res.pvalue:.3e}")
    return pvals


def fmt_p(p):
    stars = "****" if p < 1e-4 else "***" if p < 1e-3 else "**" if p < 1e-2 else "*" if p < 0.05 else "ns"
    p_txt = "p<1e-300" if p == 0 else f"p={p:.1e}"
    return stars, p_txt


exp_p = xgb_wilcoxon(exp_pcc, "experiment")
gene_p = xgb_wilcoxon(gene_pcc, "gene")

# one axis, two boxes per method: per-gene and per-experiment PCC
from matplotlib.patches import Patch

plt.style.use("default")
INK, MUTED, GRID = "#1a1a1a", "#6b6b6b", "#e6e6e3"
level_style = {
    "Per gene":       dict(pcc=gene_pcc, pvals=gene_p, color="#2a78d6", dark="#184f95", offset=-0.15),
    "Per experiment": dict(pcc=exp_pcc,  pvals=exp_p,  color="#eb6834", dark="#a8401a", offset=+0.15),
}
methods = list(gene_pcc.keys())
centers = np.arange(len(methods)) * 0.8
box_w = 0.26

fig, ax = plt.subplots(figsize=(12, 7))
np.random.seed(42)
label_y = max(np.nanmax(np.asarray(v, dtype=float)) for s in level_style.values() for v in s["pcc"].values()) + 0.06

for level, s in level_style.items():
    vals = [np.asarray(s["pcc"][m], dtype=float) for m in methods]
    vals = [v[~np.isnan(v)] for v in vals]
    pos = centers + s["offset"]

    bp = ax.boxplot(
        vals, positions=pos, widths=box_w, patch_artist=True, showfliers=False,
        showmeans=True,
        boxprops=dict(facecolor=s["color"], edgecolor=s["dark"], linewidth=1.2, alpha=0.35),
        medianprops=dict(color=INK, linewidth=2),
        whiskerprops=dict(color=s["dark"], linewidth=1.2),
        capprops=dict(color=s["dark"], linewidth=1.2),
        meanprops=dict(marker="D", markerfacecolor="white", markeredgecolor=INK, markersize=8),
        zorder=2,
    )
    # keep median and mean visible above the scatter
    for line in bp["medians"] + bp["means"]:
        line.set_zorder(4)

    # jittered scatter points
    for x, v in zip(pos, vals):
        jitter_x = np.random.normal(x, 0.02, size=len(v))
        ax.scatter(jitter_x, v, color=s["color"], edgecolors="#222222",
                   alpha=0.55, s=18, linewidth=0.4, zorder=3)

    # mean PCC on each box, just below the mean marker
    for x, v in zip(pos, vals):
        ax.annotate(f"{v.mean():.2f}", xy=(x, v.mean()), xytext=(0, -12), textcoords="offset points",
                    ha="center", va="top", fontsize=12, fontweight="bold", color=INK, zorder=6,
                    bbox=dict(boxstyle="round,pad=0.2", facecolor="white", edgecolor="none", alpha=0.85))

    # p-value of XGB vs this method at the same level, above each column
    for x, m in zip(pos, methods):
        if m == "XGBoost":
            ax.text(x, label_y, "ref", ha="center", va="bottom", fontsize=12, color=MUTED, style="italic")
            continue
        stars, p_txt = fmt_p(s["pvals"][m])
        ax.text(x, label_y, f"{stars}", ha="center", va="bottom",
                fontsize=11, color=INK, linespacing=1.1)

ax.set_xticks(centers)
ax.set_xticklabels(methods, fontsize=16, color=INK)
#ax.set_xlim(centers[0] - 0.5, centers[-1] + 0.5)
ax.set_xlim(centers[0] - 0.42, centers[-1] + 0.42)

ax.set_ylim(top=label_y + 0.2)
ax.set_ylabel("Pearson correlation (PCC)", fontsize=16, color=INK)
ax.tick_params(axis="y", labelsize=14, colors=MUTED, length=0)
ax.tick_params(axis="x", length=0)
ax.yaxis.grid(True, color=GRID, linewidth=0.8)
ax.set_axisbelow(True)
for sp in ["top", "right", "left"]:
    ax.spines[sp].set_visible(False)
ax.spines["bottom"].set_color("#bdbdb8")

ax.legend(
    handles=[Patch(facecolor=s["color"], edgecolor=s["dark"], alpha=0.6, label=l) for l, s in level_style.items()]
            + [plt.Line2D([], [], marker="D", ls="", markersize=8, markerfacecolor="white", markeredgecolor=INK,
                          label="Mean (value below)"),
               plt.Line2D([], [], color=INK, lw=2, label="Median")],
    loc="upper left", bbox_to_anchor=(0, -0.07), ncol=4, frameon=False, fontsize=14,
)
fig.text(0.99, 0.0, "**** p < 0.0001 (two-sided paired Wilcoxon signed-rank test vs. XGB)",
         ha="right", fontsize=12, color=MUTED)

plt.tight_layout()
plt.savefig("pcc_boxplot_gene_vs_experiment_wilcoxon.png", dpi=600, bbox_inches="tight", facecolor="white")
plt.show()
```

```python
from matplotlib.lines import Line2D

plt.style.use("default")
INK, MUTED, GRID = "#252525", "#666666", "#E8E8E8"
panels = [
    dict(title="Per-gene PCC", pcc=gene_pcc, pvals=gene_p, color="#4C78A8", dark="#28517D"),
    dict(title="Per-experiment PCC", pcc=exp_pcc, pvals=exp_p, color="#E68A55", dark="#A64E25"),
]
methods = list(gene_pcc.keys())
positions = np.arange(len(methods))
MAX_JITTER_POINTS = 1000  # set to None to show every observation

fig, axes = plt.subplots(1, 2, figsize=(14, 5.8), sharey=False)
rng = np.random.default_rng(42)

for ax, panel in zip(axes, panels):
    color, dark = panel["color"], panel["dark"]
    values = [np.asarray(panel["pcc"][m], dtype=float) for m in methods]
    values = [v[np.isfinite(v)] for v in values]

    # panel-specific y-axis limits
    all_vals = np.concatenate(values)
    ymin, ymax = all_vals.min(), all_vals.max()
    yrange = max(ymax - ymin, 0.1)
    p_y = ymax + 0.075 * yrange

    # jittered points (downsampled for display only)
    for x, v in zip(positions, values):
        plot_v = v
        if MAX_JITTER_POINTS is not None and len(v) > MAX_JITTER_POINTS:
            plot_v = rng.choice(v, size=MAX_JITTER_POINTS, replace=False)
        ax.scatter(rng.normal(loc=x, scale=0.075, size=len(plot_v)), plot_v, s=9, color=color,
                   alpha=0.20, edgecolors="none", zorder=1, rasterized=True)

    bp = ax.boxplot(
        values, positions=positions, widths=0.52, patch_artist=True, showfliers=False, showmeans=True,
        boxprops=dict(facecolor=color, edgecolor=dark, linewidth=1.2, alpha=0.30),
        medianprops=dict(color=INK, linewidth=1.8),
        whiskerprops=dict(color=dark, linewidth=1.2),
        capprops=dict(color=dark, linewidth=1.2),
        meanprops=dict(marker="D", markerfacecolor="white", markeredgecolor=INK, markeredgewidth=1.1, markersize=6),
        zorder=3,
    )
    for artist in bp["means"] + bp["medians"]:
        artist.set_zorder(5)

    # mean PCC labels and Wilcoxon significance vs XGB
    for x, v, method in zip(positions, values, methods):
        ax.annotate(f"{np.mean(v):.2f}", xy=(x, np.mean(v)), xytext=(0, -13), textcoords="offset points",
                    ha="center", va="top", fontsize=10, fontweight="semibold", color=INK, zorder=6,
                    bbox=dict(boxstyle="round,pad=0.18", facecolor="white", edgecolor="none", alpha=0.85))
        if method != "XGB":
            ax.text(x, p_y, fmt_p(panel["pvals"][method])[0], ha="center", va="center", fontsize=13, color=INK)

    ax.set_title(panel["title"], fontsize=15, fontweight="semibold", pad=18, color=INK)
    ax.set_xticks(positions)
    ax.set_xticklabels(methods, fontsize=11)
    ax.set_xlim(-0.6, len(methods) - 0.4)
    ax.set_ylim(ymin - 0.05 * yrange, ymax + 0.19 * yrange)
    ax.tick_params(axis="both", length=0, labelsize=10, colors=INK)
    ax.yaxis.grid(True, color=GRID, linewidth=0.8)
    ax.set_axisbelow(True)
    for spine in ["top", "right", "left"]:
        ax.spines[spine].set_visible(False)
    ax.spines["bottom"].set_color("#CCCCCC")

axes[0].set_ylabel("Pearson correlation (PCC)", fontsize=12, color=INK)

legend_handles = [
    Line2D([], [], marker="D", linestyle="", markerfacecolor="white", markeredgecolor=INK, markersize=6,
           label="Mean (number below)"),
    Line2D([], [], color=INK, linewidth=1.8, label="Median"),
]
fig.legend(handles=legend_handles, loc="lower center", bbox_to_anchor=(0.5, 0.055), ncol=2, frameon=False, fontsize=10)
fig.text(0.5, 0.015, "**** p < 0.0001 (two-sided paired Wilcoxon signed-rank test vs. XGB)",
         ha="center", fontsize=10, color=MUTED)
fig.subplots_adjust(left=0.07, right=0.98, top=0.86, bottom=0.20, wspace=0.12)

plt.savefig("pcc_gene_experiment_side_by_side.png", dpi=600, bbox_inches="tight", facecolor="white")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

matrix = np.random.randint(1, 6, size=(4, 4))

cmap = ListedColormap([
    "#E5E1DD",  # 1
    "#CDB9A8",  # 2
    "#A8755D",  # 3
    "#713F3F",    # 4
    "#542F2F"  # 5
])

fig, ax = plt.subplots(figsize=(9, 6))

norm = BoundaryNorm(
    boundaries=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5],
    ncolors=cmap.N
)

im = ax.pcolormesh(
    matrix,
    cmap=cmap,
    norm = norm,
    edgecolors="#FFFFFF",  # lines between cells
    linewidth=1.5
)



ax.set_xlim(0, 4)
ax.set_ylim(4, 0)
ax.set_xticks([])
ax.set_yticks([])
cbar = plt.colorbar(im, ax=ax,aspect=15,  ticks=[1, 2, 3, 4, 5])
cbar.ax.set_yticklabels([
    "1: Strong non-binding",
    "2: Slight non-binding",
    "3: Slight binding",
    "4: Moderate binding",
    "5: Strong binding",
], fontsize=20)
plt.tight_layout()
plt.savefig("binding_strength_heatmap.png", dpi=600, bbox_inches="tight")
plt.show()
```
